```python
import pandas as pd
import numpy as np
```

```python
pd.set_option('display.max_columns',None)
df = pd.read_excel(r"C:\Users\HPW\Downloads\ODC (Australia) Deliveries Data.xlsx")
df['Wides'] = df['Wides'].fillna(0)
df['No Balls'] = df['No Balls'].fillna(0)
```

```python
df
```

```python
df['Dot']= df['Runs Off Bat'].apply(lambda x: 1 if x == 0 else 0)
df['One']= df['Runs Off Bat'].apply(lambda x: 1 if x ==1 else 0 )
df['Two']= df['Runs Off Bat'].apply(lambda x: 1 if x ==2 else 0)
df['Three']= df['Runs Off Bat'].apply(lambda x: 1 if x == 3 else 0)

df['Four']=(
    (df['Runs Off Bat']==4) &
    (df['Non Boundary'].isna())
).astype(int)


df['Six']=(
    (df['Runs Off Bat']==6) &
    (df['Non Boundary'].isna())
).astype(int)
```

```python
balls=pd.DataFrame(df[df['Innings'].isin([1,2]) & (df['Wides']== 0 )].groupby(['Striker'])['Match ID'].count().reset_index().rename(columns ={'Striker':'Batter','Match ID':'Balls'}))
```

```python
player_stats=runs.merge(balls, on = 'Batter').merge(dot_balls, on = 'Batter').merge(innings, on = 'Batter').merge(fours, on = 'Batter').merge(sixes, on = 'Batter').merge(dismissals, on ='Batter')
```

```python
player_stats['Batting Average']= np.where(
    player_stats['Dismissals']> 0,
    player_stats['Runs'] / player_stats['Dismissals'],0 
).round(2)
```

```python
pd.DataFrame(player_stats['Batting Average'])
```

```python
player_stats['Strike Rate']=(player_stats['Runs']/player_stats['Balls']*100).round(2)
```

```python
player_stats['Runs Per Innings'] = (player_stats['Runs'] / player_stats['Innings']).round(2)
```

```python
player_stats['Ball Per Boundary']=( player_stats['Balls'] / player_stats['Fours'] + player_stats['Sixes']).round(2)
```
